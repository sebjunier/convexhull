# Calcul de l'enveloppe convexe du système Fe-Cu-Co-Ni 
Données extraites de Materials Project

```python
# Paquet necessaire : numpy | pandas | scipy
# programme py necessaire : cvh.py | cvh_fonc.py
import numpy as np
import pandas as pd
from cvh import cvhull
```

```python
df=pd.read_csv("data_test/data3.csv")
df
```

```python
nom=[x.replace(' ','') for x in df['id']]
Fe=list(df['Fe'])
Cu=list(df['Cu'])
Co=list(df['Co'])
Ni=list(df['Ni'])
tot=list(df['somme'])



compo=[[Fe[i]/tot[i],Cu[i]/tot[i],Co[i]/tot[i],Ni[i]/tot[i]] for i in range(83)]
Ef=list(df['E'])
```

```python
hull=cvhull(nom,compo,Ef)
```

## Listes des pahses stables

```python
print('La base contient {} composés dont {} stables'.format(hull.npoints,len(hull.stable)))
print('Listes des phases stables:')
print('--------------------------')
print('{:15}  Composition'.format('Phases'))
for i in hull.stable:
    txt='{:15}'.format(hull.nom[i])
    for j in range(hull.dim):
        txt+=' {:5.2f}'.format(hull.composition[i,j])
    print(txt)
```
